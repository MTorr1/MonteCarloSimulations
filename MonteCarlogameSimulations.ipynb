```python

# coin flip game


import random

sims = 5000
simcount = 0
heads = 0
percent = []

for i in range(sims) :
    simcount += 1
    random1 = random.randint(0,1)
    if random1 >0:
        heads += 1
    percent_heads = round(heads/simcount, 4)
    percent.append(percent_heads)
    print(percent_heads)

print("simulation compleate")

```

```python


# Graphing Coin Flips

import matplotlib.pyplot as plt


x = range(1, simcount+1)
y = percent

ax = plt.subplot()
ax.plot(x,y)
# ax.set_xlim(left=0)
# ax.set_ylim(bottom=0)

ax.set(xlabel='Simulations', ylabel='% Heads', title="Simulating Flipping a Coin For Heads")
```

```python
# Game 2: Dice Game - Lucky 7

import random

sims, wins, losses = 100,0,0
start_balance =50
end_balance = 0
bet_amount = 10
payoff = 6

for i in range(sims) :
    d1 = random.randint(1,6)
    d2 = random.randint(1,6)
    rollsum = d1 + d2

    if rollsum == 7 :
        wins += 1
        end_balance = start_balance + (bet_amount*payoff)
    else:
        losses += 1
        end_balance = start_balance - bet_amount
    start_balance = end_balance

print("Wins: ", wins)
print("Losses: ", losses)
print("Total rolls", sims)
print("Start Balance",start_balance)
print("End Balance", end_balance)


#!!!
#!!!
#!!!
# I would like to map the different expected profits at different bet amounts and payoffs
#!!!
#!!!
#!!!
```

```python
#Game 2 Part 2 - Pass
# Roll two dice
# win on a 7 or 11
# loss on a 2, 3 or 12
# draw on any remaining number
import random

sims, wins, losses, draws = 99,0,0,0
start_balance = 50
end_balance = 0
bet = 10
payoff = 5

for i in range(sims) :
    d1 = random.randint(1, 6)
    d2 = random.randint(1, 6)
    roll_sum = d1 + d2

    if roll_sum == 7 or roll_sum == 11 :
        wins += 1
        end_balance = start_balance + (bet*payoff)
    elif roll_sum in (2,3,12) :
        losses += 1
        end_balance = start_balance - bet
    else:
        draws += 1
        start_balance = end_balance
    start_balance = end_balance

print("Wins: ", wins)
print("Losses: ", losses)
print("Draws:", draws)
print("Total rolls", sims)
print("Start Balance",start_balance)
print("End Balance", end_balance)

#!!!
#!!!
#!!!
# I would like to map the different expected profits at different bet amounts and payoffs
#!!!
#!!!
#!!!
```

```python
# Simulation 3 - Stock Market returns simulation

import pandas as pd
import numpy as np
import random

simulations = 1000
years = 40
beginingBalance = 10000
endBalance = []

mean = .1188
sd = .1976

for i in range(simulations) :
    stockMarketSims = np.random.normal(mean,sd,years)
    bal = beginingBalance
    for j in range(years) : # Inner loop is used to apply each years return to the curent balance
        bal = bal * (1 + stockMarketSims[j])
    endBalance.append(round(bal, 2))

# print("The end Balance is: ", endBalance)



import matplotlib.pyplot as plt
import seaborn as sns
sns.histplot(endBalance, binwidth=500000)
plt.xlabel('Ending Balance')
plt.show()



# Summary Satstics

meanEndBalance = np.round(sum(endBalance) / simulations,2)
mediumEndBalance = np.round(endBalance[499],2)
sdEndBalance = np.std(endBalance, ddof=0)
roundedSDEndbalance = np.round(sdEndBalance)
fithPercentile = np.round(endBalance[49],2)
nintyFithPercentile = np.round(endBalance[949],2)

#Q: when I change the number of simulations, how do I automatically change the percentile pos

print("The Mean Endbalance is ", meanEndBalance)
print("The Medium EndBalance is : ",mediumEndBalance)
print("The Standard Devation of the possible endbalance is: ", roundedSDEndbalance) # Am I miss understanding what the std is for this data set?
print("The 5th eprcentile is :", fithPercentile)
print("The 95th Percentile is :", nintyFithPercentile)
```

```python
# Simulation 3 - Stock Market Retyurns.

# Additional Asset Classes
 
# As well as previouse code I need the historical Mean and Sd of differnt asset classdes (Gold and T Bills)

# I need to decidde my asset split then diostrubtion inital balance accordingly

# After each year, I need to find a way to rebalance portfolio to the asset split 
# - Sell assets over performing (at current price) buy assets under perfoming (at current price)


simulations = 1000 # if simulations change need to update the percentile calculations

years = 40

initalBal = 10000

stockBal = 0
goldBal = 0 
bondBal = 0

endBalance = []


stockSplit, goldSplit, bondSplit = .7, .2, .1
stockMean, goldMean, bondMean    = .1188, 0.0313, 0.030
stockSD, goldSD, bondSd          = .1976, .1942, .1070

#Arrays to follow yearly returns
###np.zeros, creates an array filled with 0s
stockHistory = np.zeros((simulations, years +1))
goldHistory = np.zeros((simulations, years +1))
bondHistory = np.zeros((simulations, years +1))
endOfYearHistory = np.zeros((simulations, years +1))



for i in range(simulations) :
    totalBal = initalBal
    stockMarketReturns = np.random.normal(stockMean, stockSD, years)
    stockBal = initalBal * stockSplit
    stockHistory[i,0] = stockBal

    goldReturns = np.random.normal(goldMean, goldSD, years)
    goldBal = initalBal * goldSplit
    goldHistory[i,0] = goldBal
    
    bondMarketReturns = np.random.normal(bondMean, bondSd, years)
    bondBal = initalBal * bondSplit
    bondHistory[i,0] = bondBal

    endOfYearHistory[i,0] = totalBal

    for j in range(years) :
        stockBal = totalBal * stockSplit * (1 + stockMarketReturns[j])
        stockHistory[i, j + 1] = stockBal

        goldBal = totalBal * goldSplit *(1 + goldReturns[j])
        goldHistory[i, j + 1] = goldBal

        bondBal = totalBal * bondSplit * (1 + bondMarketReturns[j])
        bondHistory[i, j + 1] = bondBal

        totalBal = stockBal + goldBal + bondBal
        endOfYearHistory[i, j + 1] = totalBal

    endBalance.append(round(totalBal, 2))


meanEndBalance = np.round(sum(endBalance) / simulations,2)
mediumEndBalance = np.round(endBalance[499],2)
sdEndBalance = np.std(endBalance, ddof=0)
roundedSDEndbalance = np.round(sdEndBalance)
fithPercentile = np.round(endBalance[49],2)
nintyFithPercentile = np.round(endBalance[949],2)

print("The Mean Endbalance is ", meanEndBalance)
print("The Medium EndBalance is : ",mediumEndBalance)
print("The Standard Devation of the possible endbalance is: ", roundedSDEndbalance) # Am I miss understanding what the std is for this data set?
print("The 5th eprcentile is :", fithPercentile)
print("The 95th Percentile is :", nintyFithPercentile)

### TODO: Generate graph
### Can I generate a line graph opf each asset?
### If possible, get the assets correctly corolated
### Compeae pure stock to mixed balance
# Look into AB Testing 
### Write tests

### Histogram - Shows the frequence an end balance was achieved across all simulations
import matplotlib.pyplot as plt
import seaborn as sns
sns.histplot(endBalance, binwidth=50000)
plt

### Stackplot

# The mean value for each asset at each year across all sims
# 0 tells numpy which direction to average. zero is down ( the column)
meanStockHistory = np.mean(stockHistory, 0)
meanGoldHistory = np.mean(goldHistory, 0)
meanBondHistory = np.mean(bondHistory, 0)
meanBalanceHistory = np.mean(endOfYearHistory, 0)


#Create the Graph
fig, ax = plt.subplots()

# x axis Years:
yearNumbers = np.arange(years +1)

# plot each assest
ax.plot(yearNumbers, meanStockHistory, label="Stocks")
ax.plot(yearNumbers, meanGoldHistory, label="Gold")
ax.plot(yearNumbers, meanBondHistory, label="Bonds")
ax.plot(yearNumbers, meanBalanceHistory, label="Portfolio Balance")

#Labels
ax.set_xlabel("Years")
ax.set_ylabel("Mean Balance")

#Shown Which Line
ax.legend()

plt.show()
```

```python
# Simulation 4 - My Version of Stock Market Retrurns

# Cleaning Data



# TODO: Look at distribution of S&P 500 going back to 1926

# Caluclate ARR (anual retrun rate)

# Find mean and SD

# Generate Frequency distribution of historical returns - It shouild be skewed but approximently normal distrubtion

# Track a hypothetical retierment portfolio over 40 years (Loojk aat spreadsheet model and Holman 2019)




import numpy as np

import pandas as pd

import matplotlib



inital_investment = 10000



df = pd.read_csv('/Users/matthewtorr/Documents/VSCode/Personal Projects/Maths/Data/Shillers_S&P_Data.csv')

#converts coluum to datetime
df['Date'] = pd.to_datetime(df['Date']) 

#removes all rows where date is < 1926-01-01
df_filtered = df[df['Date'].between('1926-01-01','2023-01-01')] 

df_filtered['Capital Gain Yeild'] = (df_filtered['Real Price'] - df_filtered['Real Price'].shift(1)) / df_filtered['Real Price'].shift(1)
df_filtered['Dividend Yeild'] = (df_filtered['Real Dividend']/12) / df_filtered['Real Price'].shift(1)
df_filtered['Total Monthly Return'] = df_filtered['Capital Gain Yeild'] + df_filtered['Dividend Yeild']
df_filtered['Rolling Balance'] = np.round(inital_investment * (1 + df_filtered['Total Monthly Return']).cumprod(),2)


df_filtered['Percent_change_since_last_month'] = (df_filtered['Rolling Balance'].pct_change() * 100).round(2)
df_filtered['Percent_change_since_last_year'] = (df_filtered['Rolling Balance'].pct_change(12) * 100).round(2)
df_filtered.to_csv('S&P_Data.csv', index=False)
```
